```python
import pandas as pd

file_path = r"C:\Users\Dell\OneDrive\Documents\Project 1\ecommerce_portfolio_dataset_100k.csv"

df = pd.read_csv(file_path)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
print(df.head())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nUnique Values:")
print(df.nunique())
```

```python
# STEP 2: DATA QUALITY AUDIT

print("=" * 60)
print("DATA QUALITY AUDIT")
print("=" * 60)

# 1. Duplicate Order IDs
print("\n1. Duplicate Order IDs:")
print(df["order_ID"].duplicated().sum())

# 2. Show duplicate Order IDs
print("\n2. Sample Duplicate Order IDs:")
print(df[df["order_ID"].duplicated(keep=False)]
      .sort_values("order_ID")
      .head(10))

# 3. Missing values percentage
print("\n3. Missing Value Percentage:")
missing_percentage = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
print(missing_percentage[missing_percentage > 0])

# 4. Data types
print("\n4. Data Types:")
print(df.dtypes)

# 5. Numeric summary
print("\n5. Numeric Summary:")
print(df.describe())

# 6. Categorical values
print("\n6. Order Status:")
print(df["order_status"].value_counts())

print("\n7. Return Flag:")
print(df["return_flag"].value_counts())

print("\n8. Customer Type:")
print(df["customer_type"].value_counts())

print("\n9. Payment Method:")
print(df["payment_method"].value_counts())

print("\n10. Product Category:")
print(df["product_category"].value_counts())
```

```python
# STEP 3: BUSINESS LOGIC VALIDATION

print("=" * 60)
print("BUSINESS LOGIC VALIDATION")
print("=" * 60)


# 1. Order Status vs Return Flag
print("\n1. Order Status vs Return Flag:")
print(pd.crosstab(df["order_status"], df["return_flag"]))


# 2. Return Flag vs Return Reason
print("\n2. Return Flag vs Return Reason:")

return_reason_check = pd.crosstab(
    df["return_flag"],
    df["return_reason"].isna()
)

print(return_reason_check)


# 3. Coupon Used vs Coupon Code
print("\n3. Coupon Used vs Coupon Code:")

coupon_check = pd.crosstab(
    df["coupon_used"],
    df["coupon_code"].isna()
)

print(coupon_check)


# 4. Customer Type vs Previous Orders
print("\n4. Customer Type vs Previous Orders:")

print(
    df.groupby("customer_type")["previous_orders"]
    .agg(["min", "max", "mean"])
)


# 5. Delivery Days vs Order Status
print("\n5. Delivery Days vs Order Status:")

print(
    df.groupby("order_status")["delivery_days"]
    .agg(["count", "min", "max", "mean"])
)


# 6. Customer Rating vs Order Status
print("\n6. Customer Rating vs Order Status:")

print(
    df.groupby("order_status")["customer_rating"]
    .agg(["count", "min", "max", "mean"])
)


# 7. Check impossible ages
print("\n7. Customer Age Check:")

print(
    df[(df["customer_age"] < 18) | (df["customer_age"] > 100)]
)


# 8. Check impossible quantities
print("\n8. Quantity Check:")

print(
    df[(df["quantity"] <= 0)]
)


# 9. Check negative prices/order amounts
print("\n9. Negative Price / Amount Check:")

print(
    df[
        (df["unit_price"] <= 0) |
        (df["order_amount"] <= 0)
    ]
)


# 10. Check return reason for non-returned orders
print("\n10. Return Reason on Non-Returned Orders:")

print(
    df[
        (df["return_flag"] == 0) &
        (df["return_reason"].notna())
    ][
        ["order_ID", "order_status", "return_flag", "return_reason"]
    ].head(10)
)


# 11. Check coupon code when coupon was not used
print("\n11. Coupon Code when Coupon was NOT Used:")

print(
    df[
        (df["coupon_used"] == "No") &
        (df["coupon_code"].notna())
    ][
        ["order_ID", "coupon_used", "coupon_code"]
    ].head(10)
)
```

```python
# STEP 4.1 - REMOVE EXACT DUPLICATES

print("Rows before removing duplicates:", len(df))

df_clean = df.drop_duplicates().copy()

print("Rows after removing duplicates:", len(df_clean))

print("Duplicates removed:", len(df) - len(df_clean))
```

```python
# STEP 4.2 - CONVERT DATE AND TIME

df_clean["order_date"] = pd.to_datetime(
    df_clean["order_date"],
    errors="coerce"
)

df_clean["order_time"] = pd.to_datetime(
    df_clean["order_time"],
    format="%H:%M:%S",
    errors="coerce"
).dt.time

print(df_clean[["order_date", "order_time"]].head())

print("\nData Types:")
print(df_clean[["order_date", "order_time"]].dtypes)
```

```python
# STEP 4.3 - HANDLE RETURN REASON

df_clean["return_reason"] = df_clean["return_reason"].fillna(
    "Not_Returned"
)
```

```python
# STEP 4.4 - HANDLE COUPON CODE

df_clean["coupon_code"] = df_clean["coupon_code"].fillna(
    "No_Coupon"
)
```

```python
df_clean["gross_amount"] = (
    df_clean["quantity"] *
    df_clean["unit_price"]
)
```

```python
df_clean["discount_amount"] = (
    df_clean["gross_amount"] *
    df_clean["discount_percentage"] / 100
)
```

```python
df_clean["calculated_net_amount"] = (
    df_clean["gross_amount"] -
    df_clean["discount_amount"]
)
```

```python
# STEP 4.8 - VALIDATE ORDER AMOUNT

df_clean["amount_difference"] = (
    df_clean["order_amount"] -
    df_clean["calculated_net_amount"]
)

print(
    df_clean["amount_difference"].describe()
)
```

```python
print(
    "Rows with amount mismatch:",
    (df_clean["amount_difference"].abs() > 1).sum()
)
```

```python
# STEP 4.10 - SAVE CLEANED DATASET

cleaned_file_path = "ecommerce_cleaned_data.csv"

df_clean.to_csv(
    cleaned_file_path,
    index=False
)

print("Cleaned dataset saved successfully!")
print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))
```

```python
df_clean["order_date"] = pd.to_datetime(
    df_clean["order_date"],
    errors="coerce"
)

print("Data type:", df_clean["order_date"].dtype)
print("Invalid dates:", df_clean["order_date"].isna().sum())
```

```python
# Check existing columns
cols = df_clean.columns.tolist()

# 1. Gross amount
if "unit_price" in cols and "quantity" in cols:
    df_clean["gross_amount"] = df_clean["unit_price"] * df_clean["quantity"]

# 2. Discount Amount (Auto-detecting available column)
if "discount_percent" in cols:
    df_clean["discount_amount"] = df_clean["gross_amount"] * (df_clean["discount_percent"] / 100)
elif "discount" in cols:
    df_clean["discount_amount"] = df_clean["gross_amount"] * (df_clean["discount"] / 100)
elif "discount_rate" in cols:
    df_clean["discount_amount"] = df_clean["gross_amount"] * (df_clean["discount_rate"] / 100)
else:
    # Agar koi discount column nahi hai toh discount 0 maan lega
    df_clean["discount_amount"] = 0

# 3. Calculated Net Amount
df_clean["calculated_net_amount"] = df_clean["gross_amount"] - df_clean["discount_amount"]

# 4. STEP 4.5 - VALIDATE ORDER AMOUNT
if "order_amount" in cols:
    df_clean["amount_difference"] = (
        df_clean["order_amount"] - 
        df_clean["calculated_net_amount"]
    )
    print("Rows with amount mismatch:", (df_clean["amount_difference"].abs() > 1).sum())
    print("Maximum difference:", df_clean["amount_difference"].abs().max())
```

```python
# Missing percentage

missing_percentage = (
    df_clean.isnull().mean() * 100
).round(2)

print("\nMissing Percentage:")
print(missing_percentage[missing_percentage > 0])
```

```python
# STEP 4.7 - HANDLE DELIVERY DAYS AND CUSTOMER RATING

# Create availability flags
df_clean["delivery_days_available"] = (
    df_clean["delivery_days"].notna().astype(int)
)

df_clean["rating_available"] = (
    df_clean["customer_rating"].notna().astype(int)
)

# Keep the original missing values
print("Missing delivery_days:",
      df_clean["delivery_days"].isna().sum())

print("Missing customer_rating:",
      df_clean["customer_rating"].isna().sum())

print("\nDelivery days available:")
print(df_clean["delivery_days_available"].value_counts())

print("\nRating available:")
print(df_clean["rating_available"].value_counts())
```

```python
# STEP 4.8 - FINAL MISSING VALUE CHECK

print("Final Missing Value Summary:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```
